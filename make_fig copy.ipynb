```python
from datagen import make_decks, get_next_seed, save_decks
from processing import iterate_all_combos, iterate_all_combos_for_points
from make_fig import (
    load_data,
    process_data,
    create_heatmap_data,
    plot_heatmap
)
```

```python
def main(n_decks, strategy):

    # standardize user input
    strategy = strategy.strip().lower()

    if strategy not in ["tricks", "cards"]:
        raise ValueError("Strategy must be either 'Tricks' or 'Cards'.")

    # 1. Generate decks

    seed = get_next_seed()

    decks = make_decks(
        seed=seed,
        n_decks=n_decks,
        n_cards=52
    )

    deck_file = save_decks(
        decks=decks,
        seed=seed
    )

    print(f"Created {n_decks} decks.")
    print(f"Saved decks to {deck_file}")

    # -----------------------------
    # 2. Process decks
    # -----------------------------

    if strategy == "tricks":

        iterate_all_combos(
            deck_files=[deck_file]
        )

        db_path = "tricks.db"
        table_name = "trick_stats"

    elif strategy == "cards":

        iterate_all_combos_for_points(
            deck_files=[deck_file]
        )

        db_path = "points.db"
        table_name = "combo_stats"

    # -----------------------------
    # 3. Load results
    # -----------------------------

    data = load_data(
        db_path,
        table_name
    )

    data = process_data(data)

    # -----------------------------
    # 4. Prepare heatmap
    # -----------------------------

    heatmap_data, label_data = create_heatmap_data(data)

    # -----------------------------
    # 5. Plot
    # -----------------------------

    plot_heatmap(
        heatmap_data,
        label_data,
        n_decks=n_decks,
        strategy=strategy
    )


if __name__ == "__main__":

    n_decks = int(
        input("How many decks would you like to simulate? ")
    )

    strategy = input(
        "Which strategy would you like to use (Tricks or Cards)? "
    )

    main(
        n_decks=n_decks,
        strategy=strategy
    )
```
